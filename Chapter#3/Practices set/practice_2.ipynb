## *Write a Program to that entered the name and date to fill the letter format.*
```
Letter = '''
Dear <|Name|>
You are selected!
<|Date|>
'''
```

## ***Solution:***

#### ***By using replace function:***

````python
Letter = '''
Dear <|Name|>
You are selected!
<|Date|>
'''
print(Letter.replace("<|Name|>", "taha").replace("<|Date|>", "25 May 2025"))
````
