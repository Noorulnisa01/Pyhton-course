# ***Practice 1:***
## *Write a Program to display a user entered named followed by good afternoon using input function.*

## ***Solution:***

#### ***Method 1: String concatenation***

```python
name = input("Enter your name")
print("Good Afternoon " + name)
```

#### ***Method 2: f-string***
*1. f-string is used to format the string.* <br>
*2. f-strings allow you to insert variables directly into your strings in a clean and readable way.*

```python
name = input("Enter your name again")
print(f"Good Afternoon {name}")
```
