# ***Practice 3:***
## *Write a Program to detect a double space in a string.*

## ***Solution:***

#### ***By using find function:***

```python
name = "My name is   batool."
print(name.find("   "))
```

# ***Practice 4:***
## *Replace the 3 space with single space in a string.*

## ***Solution:***

#### ***By using replace function:***

```python
name = "My name is   batool."
print(name.replace("   ", " "))
```
